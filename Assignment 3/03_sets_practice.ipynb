Q22. Set Operations
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}
Find union, intersection, a - b, and symmetric difference.

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7, 8}

print("Union:", a | b)
print("Intersection:", a & b)
print("a - b:", a - b)
print("Symmetric Difference:", a ^ b)
```

Q27. Three-Set Analysis
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}
Find students in all three, exactly Python and SQL (not AWS), and at least two.

```python
python = {"Ali", "Sara", "Ahmed", "Zain"}
sql = {"Sara", "Ahmed", "Hamza"}
aws = {"Ahmed", "Zain", "Hamza"}

print("All three:", python & sql & aws)
print("Python and Sql not AWS:", (python & sql) - aws)
print("At least two:", (python & sql) | (python & aws) | (sql & aws))
```

Q28. Transaction Analysis
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}
Find IDs appearing on all days, exactly one day, and at least two days.

```python
day1 = {101, 102, 103, 104}
day2 = {103, 104, 105, 106}
day3 = {104, 106, 107, 108}

print("All days:", day1 & day2 & day3)
print("Exactly one day:", (day1 | day2 | day3) - ((day1 & day2) | (day1 & day3) | (day2 & day3)))
print("At least two days:", (day1 & day2) | (day1 & day3) | (day2 & day3))
```

Q29. Access Control
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}
Find missing and extra permissions and test full authorization

```python
required_permissions = {"read", "write", "execute"}
user_permissions = {"read", "write", "delete"}

print("Missing:", required_permissions - user_permissions)
print("Extra:", user_permissions - required_permissions)
print("Fully Authorized:", required_permissions.issubset(user_permissions))
```

Q30. Customer Segmentation
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}
Find customers on exactly one, exactly two, and all three platforms.

```python
website = {"C1", "C2", "C3", "C4", "C5"}
mobile = {"C3", "C4", "C6", "C7"}
store = {"C2", "C4", "C7", "C8"}

all_three = website & mobile & store
exactly_two = ((website & mobile) | (website & store) | (mobile & store)) - all_three
exactly_one = (website | mobile | store) - ((website & mobile) | (website & store) | (mobile & store))

print("Exactly one:", exactly_one)
print("Exactly two:", exactly_two)
print("All three:", all_three)
```
