students = [

 {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}

Calculate each average, top student, tuple of names, and set of unique marks.

Q45. Course Enrollment
courses = {
 "Python": {"Ali", "Sara", "Ahmed"},
 "SQL": {"Sara", "Ahmed", "Zain"},
 "AWS": {"Ali", "Ahmed", "Zain"}
}
Find students in exactly two, all three, and exactly one course; count unique students

```python
courses = {
 "Python": {"Ali", "Sara", "Ahmed"},
 "SQL": {"Sara", "Ahmed", "Zain"},
 "AWS": {"Ali", "Ahmed", "Zain"}
}

all_three = (courses["Python"] & courses["SQL"] & courses["AWS"])
print("All three courses:", all_three)

python_sql = courses["Python"] & courses["SQL"]
python_aws = courses["Python"] & courses["AWS"]
sql_aws = courses["SQL"] & courses["AWS"]

exactly_two = (
    (python_sql - courses["AWS"])
    | (python_aws - courses["SQL"])
    | (sql_aws - courses["Python"])
)
print("Exactly two courses:", exactly_two)

exactly_one = (
    (courses["Python"] - courses["SQL"] - courses["AWS"])
    | (courses["SQL"] - courses["Python"] - courses["AWS"])
    | (courses["AWS"] - courses["Python"] - courses["SQL"])
)
print("Exactly one course:", exactly_one)

all_students = (courses["Python"] | courses["SQL"] | courses["AWS"])
print("Unique student count:", len(all_students))
```

Q47. Pipeline Monitoring
pipelines = {
 "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
 "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
 "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}
Find total processed records, failed pipelines, largest pipeline total, and average records per run for each.

```python
pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

p1 = pipelines["pipeline_1"]
p2 = pipelines["pipeline_2"]
p3 = pipelines["pipeline_3"]

total1 = sum(p1["records"])
total2 = sum(p2["records"])
total3 = sum(p3["records"])

total_processed = total1 + total2 + total3

failed_pipelines = []

(p1["status"] == "failed") and failed_pipelines.append("pipeline_1")
(p2["status"] == "failed") and failed_pipelines.append("pipeline_2")
(p3["status"] == "failed") and failed_pipelines.append("pipeline_3")

largest_pipeline_total = max(total1, total2, total3)

average1 = total1 / len(p1["records"])
average2 = total2 / len(p2["records"])
average3 = total3 / len(p3["records"])

averages = {
    "pipeline_1": average1,
    "pipeline_2": average2,
    "pipeline_3": average3
}

print("Total processed records:", total_processed)
print("Failed pipelines:", failed_pipelines)
print("Largest pipeline total:", largest_pipeline_total)
print("Average records per run:", averages)
```

Q48. Banking Transactions
accounts = {
 "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
 "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
 "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}
Assuming zero starting balances, calculate final balances, highest balance, total withdrawals, and largest single
deposit.

```python
accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

t1 = accounts["A101"]["transactions"]
t2 = accounts["A102"]["transactions"]
t3 = accounts["A103"]["transactions"]

balance1 = sum(t1)
balance2 = sum(t2)
balance3 = sum(t3)

final_balances = {
    "A101": balance1,
    "A102": balance2,
    "A103": balance3
}

highest_balance = max(balance1, balance2, balance3)

total_withdrawals = (
    abs(min(t1[1], 0)) + abs(min(t1[3], 0))
    + abs(min(t2[1], 0)) + abs(min(t2[2], 0))
    + abs(min(t3[1], 0))
)

largest_deposit = max(
    t1[0], t1[2],
    t2[0],
    t3[0], t3[2]
)

print("Final balances:", final_balances)
print("Highest balance:", highest_balance)
print("Total withdrawals:", total_withdrawals)
print("Largest single deposit:", largest_deposit)
```

Q49. Product Recommendations
users = {
 "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
 "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
 "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}
Find products bought by at least two users, Ali wishlist minus purchases, shared wishlist products, and all products.

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

ali_purchased = users["Ali"]["purchased"]
sara_purchased = users["Sara"]["purchased"]
ahmed_purchased = users["Ahmed"]["purchased"]

ali_wishlist = users["Ali"]["wishlist"]
sara_wishlist = users["Sara"]["wishlist"]
ahmed_wishlist = users["Ahmed"]["wishlist"]

bought_by_two = (
    (ali_purchased & sara_purchased)
    | (ali_purchased & ahmed_purchased)
    | (sara_purchased & ahmed_purchased)
)

ali_wishlist_minus_purchases = ali_wishlist - ali_purchased

shared_wishlist = (
    (ali_wishlist & sara_wishlist)
    | (ali_wishlist & ahmed_wishlist)
    | (sara_wishlist & ahmed_wishlist)
)

all_products = (
    ali_purchased | sara_purchased | ahmed_purchased
    | ali_wishlist | sara_wishlist | ahmed_wishlist
)

print("Bought by at least two users:", bought_by_two)
print("Ali's wishlist minus purchases:", ali_wishlist_minus_purchases)
print("Shared wishlist products:", shared_wishlist)
print("All products:", all_products)
```

Q50. Final Company Analysis
company = {
 "departments": {
 "Engineering": {
 "employees": [
 ("Ali", 120000, {"Python", "AWS", "SQL"}),
 ("Sara", 150000, {"Python", "Databricks", "SQL"})
 ],
 "budget": 500000
 },
 "Analytics": {
 "employees": [
 ("Ahmed", 100000, {"SQL", "Power BI"}),
 ("Zain", 130000, {"Python", "SQL", "Tableau"})
 ],
 "budget": 400000
 }
 }
}
1. Payroll per department. 2. Company payroll. 3. Highest-paid employee. 4. Skills shared by all. 5. Engineering-only
skills. 6. Remaining budget per department. 7. Department with higher average salary. 8. Name-to-salary dictionary. 9. All unique skills. 10. Is combined budget sufficient?

```python
company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

departments = company["departments"]

engineering = departments["Engineering"]
analytics = departments["Analytics"]

e1 = engineering["employees"][0]
e2 = engineering["employees"][1]
a1 = analytics["employees"][0]
a2 = analytics["employees"][1]

# 1. Payroll per department
engineering_payroll = e1[1] + e2[1]
analytics_payroll = a1[1] + a2[1]

payroll_per_department = {
    "Engineering": engineering_payroll,
    "Analytics": analytics_payroll
}

# 2. Company payroll
company_payroll = engineering_payroll + analytics_payroll

# 3. Highest paid employee
highest_paid = max(e1, e2, a1, a2, key=lambda employee: employee[1])

# 4. Skills shared by all employees
shared_skills = e1[2] & e2[2] & a1[2] & a2[2]

# 5. Engineering only skills
engineering_skills = e1[2] | e2[2]
analytics_skills = a1[2] | a2[2]
engineering_only_skills = engineering_skills - analytics_skills

# 6. Remaining budget per department
engineering_remaining = engineering["budget"] - engineering_payroll
analytics_remaining = analytics["budget"] - analytics_payroll

remaining_budgets = {
    "Engineering": engineering_remaining,
    "Analytics": analytics_remaining
}

# 7. Department with higher average salary
engineering_average = engineering_payroll / len(engineering["employees"])
analytics_average = analytics_payroll / len(analytics["employees"])

higher_average_department = (
    "Engineering" if engineering_average > analytics_average
    else "Analytics" if analytics_average > engineering_average
    else "Equal"
)

# 8. Name to salary dictionary
name_to_salary = {
    e1[0]: e1[1],
    e2[0]: e2[1],
    a1[0]: a1[1],
    a2[0]: a2[1]
}

# 9. All unique skills
all_skills = e1[2] | e2[2] | a1[2] | a2[2]

# 10. Is combined budget sufficient?
combined_budget = engineering["budget"] + analytics["budget"]
budget_sufficient = combined_budget >= company_payroll

print("1. Payroll per department:", payroll_per_department)
print("2. Company payroll:", company_payroll)
print("3. Highest-paid employee:", highest_paid[0], highest_paid[1])
print("4. Skills shared by all:", shared_skills)
print("5. Engineering-only skills:", engineering_only_skills)
print("6. Remaining budgets:", remaining_budgets)
print("7. Higher average salary:", higher_average_department)
print("   Engineering average:", engineering_average)
print("   Analytics average:", analytics_average)
print("8. Name-to-salary:", name_to_salary)
print("9. All unique skills:", all_skills)
print("10. Combined budget sufficient:", budget_sufficient)
```
