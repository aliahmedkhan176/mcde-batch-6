Q35. Student Marks
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}
Find highest, lowest, average, and top-scoring student without loops or lambda

```python
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

highest = max(marks.values())
lowest = min(marks.values())
average = sum(marks.values()) / len(marks)
top_student = max(marks, key=marks.get)

print("Highest marks:", highest)
print("Lowest marks:", lowest)
print("Average marks:", average)
print("Top-scoring student:", top_student)
```

Q36. Nested Employee Records
employees = {
 "E101": {"name": "Ali", "salary": 80000},

 "E103": {"name": "Ahmed", "salary": 95000}
}
Extract Sara salary, increase Ali salary 15%, and calculate total payroll

```python
employees = {
 "E101": {"name": "Ali", "salary": 80000},
 "E102": {"name": "Sara", "salary": 120000},
 "E103": {"name": "Ahmed", "salary": 95000}
}

sara_salary = employees["E102"]["salary"]
employees["E101"]["salary"] *= 1.15
total_payroll = (employees["E101"]["salary"] + employees["E102"]["salary"] + employees["E103"]["salary"])

print("Sara's Salary:", sara_salary)
print("Ali's new Salary:", employees["E101"]["salary"])
print("Total payroll:", total_payroll)
```

Q37. Dictionary Inversion
data = {"Python": "Ali", "SQL": "Sara", "AWS": "Ahmed"}
Create inverse dictionary without loops or comprehensions.

```python
data = {"Python": "Ali", "SQL": "Sara", "AWS": "Ahmed"}

inverse_dict = dict(zip(data.values(), data.keys()))

print(inverse_dict)
```

Q38. Order Calculation
order = {
 "laptop": {"price": 120000, "quantity": 2},
 "mouse": {"price": 1500, "quantity": 5},
 "keyboard": {"price": 3500, "quantity": 3}
}
Calculate subtotals, grand total, and amount after 10% discount.

```python
order = {
 "laptop": {"price": 120000, "quantity": 2},
 "mouse": {"price": 1500, "quantity": 5},
 "keyboard": {"price": 3500, "quantity": 3}
}

laptop = order["laptop"]["price"] * order["laptop"]["quantity"]
mouse = order["mouse"]["price"] * order["mouse"]["quantity"]
keyboard = order["keyboard"]["price"] * order["keyboard"]["quantity"]

grand_total = laptop + mouse + keyboard
discounted_total = grand_total * 0.90

print("Laptop subtotal:", laptop)
print("Mouse subtotal:", mouse)
print("Keyboard subtotal:", keyboard)
print("Grand total:", grand_total)
print("After 10% discount:", discounted_total)
```

Q39. Employee Performance
employees = {
 "Ali": {"python": 90, "sql": 85, "aws": 88},
 "Sara": {"python": 95, "sql": 92, "aws": 90},
 "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}
Calculate each average, top performer, and highest SQL score.

```python
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

averages = {
    "Ali": sum(employees["Ali"].values()) / len(employees["Ali"]),
    "Sara": sum(employees["Sara"].values()) / len(employees["Sara"]),
    "Ahmed": sum(employees["Ahmed"].values()) / len(employees["Ahmed"])
}

sql_scores = {
    "Ali": employees["Ali"]["sql"],
    "Sara": employees["Sara"]["sql"],
    "Ahmed": employees["Ahmed"]["sql"]
}

print("Averages:", averages)
print("Top performer:", max(averages, key=averages.get))
print("Highest SQL score:", max(sql_scores.values()))
```

Q40. Inventory Management
inventory = {
 "P101": {"name": "Laptop", "price": 120000, "stock": 5},
 "P102": {"name": "Mouse", "price": 1500, "stock": 20},
 "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}
Calculate total value, most valuable item by price x stock, and update stock after selling two laptops and five mice.

```python
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

total_value = (
    inventory["P101"]["price"] * inventory["P101"]["stock"]
    + inventory["P102"]["price"] * inventory["P102"]["stock"]
    + inventory["P103"]["price"] * inventory["P103"]["stock"]
)

values = {
    inventory["P101"]["name"]: inventory["P101"]["price"] * inventory["P101"]["stock"],
    inventory["P102"]["name"]: inventory["P102"]["price"] * inventory["P102"]["stock"],
    inventory["P103"]["name"]: inventory["P103"]["price"] * inventory["P103"]["stock"]
}

most_valuable = max(values, key=values.get)

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

print("Total inventory value:", total_value)
print("Most valuable item:", most_valuable)
print("Updated inventory:", inventory)
```
